# Figure 5: uniform jitter frequency sweep (3, 5.5, 8, 10 Hz)

A: FFT at each jitter frequency. B: % reduction vs. frequency, jitter vs. fixed open loop, with closed loop and OL 5.5 Hz references.

```python
import cairosvg
import figurefirst as fifi
from IPython.display import display, SVG

from flutter.plotting import set_style
from flutter.config import OUTPUT_DIR
from flutter.jitter_sweep import (load_jitter_fft_data, load_psd_based_reduction_data,
                                  draw_overlaid_panel, draw_jitter_fft_row, JITTER_FREQUENCIES)
from flutter.fft_energy import add_panel_label

set_style()
```

```python
layout = fifi.svg_to_axes.FigureLayout('fig5.svg', autogenlayers=True, make_mplfigures=True, hide_layers=[])
print(layout.axes.keys())
print(layout.figures.keys())
```

```python
jitter_fft_results = load_jitter_fft_data()
jitter_data, fixed_ol_data, cl_ref, ol55_ref = load_psd_based_reduction_data(jitter_fft_results)
```

```python
draw_overlaid_panel(layout.axes[('fig5', 'Off_Freq_Jitter')], jitter_data, fixed_ol_data, cl_ref, ol55_ref, scale=1.3)

fft_axes = [
    layout.axes[('fig5', 'Off_Freq_Jitter_3HZ')],
    layout.axes[('fig5', 'Off_Freq_Jitter_55HZ')],
    layout.axes[('fig5', 'Off_Freq_Jitter_8HZ')],
    layout.axes[('fig5', 'Off_Freq_Jitter_10HZ')],
]
draw_jitter_fft_row(fft_axes, jitter_fft_results, JITTER_FREQUENCIES, scale=0.8,
                    show_legend=False, show_inset=True, inset_show_ticks=False)
```

```python
add_panel_label(fft_axes[0], 'A', x=-0.15, y=1.15)
add_panel_label(layout.axes[('fig5', 'Off_Freq_Jitter')], 'B', x=-0.035, y=1.1)
```

```python
layout.append_figure_to_layer(layout.figures['none'], 'none', cleartarget=True)
out_svg = OUTPUT_DIR / 'fig5_filled.svg'
layout.write_svg(str(out_svg))
display(SVG(str(out_svg)))
```

```python
cairosvg.svg2pdf(url=str(out_svg), write_to=str(OUTPUT_DIR / 'fig5_filled.pdf'))
```
