# Figure 2: flutter reduction vs. actuation phase offset (0–330°)

```python
import cairosvg
import figurefirst as fifi
from IPython.display import display, SVG

from flutter.plotting import set_style
from flutter.config import OUTPUT_DIR
from flutter.phase_sweep import load_phase_sweep_data, draw_violin_panel

set_style()
```

```python
layout = fifi.svg_to_axes.FigureLayout('fig2.svg', autogenlayers=True, make_mplfigures=True, hide_layers=[])
print(layout.axes.keys())
print(layout.figures.keys())
```

```python
data = load_phase_sweep_data()   # reads config.PHASE_SWEEP_DIR
```

```python
ax = layout.axes[('fig2', 'PhaseViolin')]
draw_violin_panel(ax, data['phases'], data['results'], data['reductions'],
                  data['ns'], data['mean_red'], scale=0.5,
                  show_pct_labels=False, show_legend=False, show_n_labels=False)
fifi.mpl_functions.set_fontsize(ax, 6)
```

```python
layout.append_figure_to_layer(layout.figures['none'], 'none', cleartarget=True)
out_svg = OUTPUT_DIR / 'fig2_filled.svg'
layout.write_svg(str(out_svg))
display(SVG(str(out_svg)))
```

```python
cairosvg.svg2pdf(url=str(out_svg), write_to=str(OUTPUT_DIR / 'fig2_filled.pdf'))
```
