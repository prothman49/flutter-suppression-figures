# Figure 3: off-frequency open loop (3, 5.5, 8, 10 Hz) vs. frequency-tracked closed loop

```python
import cairosvg
import figurefirst as fifi
from IPython.display import display, SVG

from flutter.plotting import set_style
from flutter.config import OUTPUT_DIR
from flutter.frequency_profile import (load_frequency_profile_data, draw_fft_row,
                                       draw_frequency_triptych_panel, FREQ_CONDITIONS)
from flutter.fft_energy import add_panel_label

set_style()
```

```python
layout = fifi.svg_to_axes.FigureLayout('fig3.svg', autogenlayers=True, make_mplfigures=True, hide_layers=[])
print(layout.axes.keys())
print(layout.figures.keys())
```

```python
all_results = load_frequency_profile_data()
```

```python
axes_fft = [
    layout.axes[('fig3', '3hz')],
    layout.axes[('fig3', '5.5hz')],
    layout.axes[('fig3', 'CL')],
    layout.axes[('fig3', '8Hz')],
    layout.axes[('fig3', '10Hz')],
]
draw_fft_row(axes_fft, all_results, FREQ_CONDITIONS, scale=0.8, show_inset=True, show_legend=False)

draw_frequency_triptych_panel(layout.axes[('fig3', 'Total_Power')], all_results, FREQ_CONDITIONS, metric='energy_total', scale=0.8)
draw_frequency_triptych_panel(layout.axes[('fig3', 'Flutter_Power')], all_results, FREQ_CONDITIONS, metric='energy_flutter', scale=0.8)
draw_frequency_triptych_panel(layout.axes[('fig3', 'Flutter_Reduct')], all_results, FREQ_CONDITIONS, metric='pct_reduction', scale=0.8)
```

```python
add_panel_label(axes_fft[0], 'A', y=1.15, x=-0.14)
add_panel_label(layout.axes[('fig3', 'Total_Power')], 'B')
add_panel_label(layout.axes[('fig3', 'Flutter_Power')], 'C')
add_panel_label(layout.axes[('fig3', 'Flutter_Reduct')], 'D')
```

```python
layout.append_figure_to_layer(layout.figures['none'], 'none', cleartarget=True)
out_svg = OUTPUT_DIR / 'fig3_filled.svg'
layout.write_svg(str(out_svg))
display(SVG(str(out_svg)))
```

```python
cairosvg.svg2pdf(url=str(out_svg), write_to=str(OUTPUT_DIR / 'fig3_filled.pdf'))
```
